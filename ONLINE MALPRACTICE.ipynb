```python
class MatrixMultiplicationSimple {
    public static void main(String[] args) {
        // First matrix
        int a[][] = { {1, 2}, {3, 4} };
        // Second matrix
        int b[][] = { {5, 6}, {7, 8} };

        // Result matrix
        int c[][] = new int[2][2];

        // Matrix multiplication logic
        for (int i = 0; i < 2; i++) {
            for (int j = 0; j < 2; j++) {
                c[i][j] = 0;
                for (int k = 0; k < 2; k++) {
                    c[i][j] += a[i][k] * b[k][j];
                }
            }
        }

        // Display result
        System.out.println("Resultant Matrix:");
        for (int i = 0; i < 2; i++) {
            for (int j = 0; j < 2; j++) {
                System.out.print(c[i][j] + " ");
            }
            System.out.println();
        }
    }
}

```
